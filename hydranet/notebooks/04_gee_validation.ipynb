# 04 — Optional: Observed Flood Extents via Earth Engine
Validates predicted extents against observed flood extents (Sentinel-1). Fill in a real event.

```
import ee, geemap
ee.Authenticate(); ee.Initialize(project='YOUR-PROJECT')
aoi = ee.Geometry.Point([-90.0, 35.0]).buffer(20000)
s1 = (ee.ImageCollection('COPERNICUS/S1_GRD')
        .filterBounds(aoi).filterDate('2023-08-01', '2023-08-20')
        .filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VV'))
        .mosaic())
water = s1.select('VV').lt(-18)   # simple threshold -> observed extent
geemap.ee_export_image(water, 'observed_extent.tif', scale=30, region=aoi)
```
